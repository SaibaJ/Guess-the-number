````python
"""Guess The Number: a simple command-line game.

The computer picks a random number and the player has a limited number of
attempts to guess it. Statistics are tracked across rounds.
"""
import random

# ---------- Game settings ----------
lower_bound = 0
upper_bound = 100
max_attempts = 6

# ---------- Global statistics counters ----------
games_played = 0
games_won = 0
games_lost = 0
total_guesses = 0   # valid guesses made across all games
best_game = None    # fewest guesses used in a win (None until the first win)

def rules():
    """Print the game rules."""
    print("Welcome to Guess The Number, let's get started.")
    print("The rules are simple:")
    print(f"Input a guess between {lower_bound} and {upper_bound}.")
    print(f"But be wary, there are limited guesses! You have {max_attempts} attempts.\n")

def get_valid_guess(attempt):
    """Keep asking until the player enters a whole number within the bounds.

    Invalid input never leaves this function, so it is never counted
    as an attempt by the caller.
    """
    while True:
        raw = input(f"Attempt {attempt}/{max_attempts} - type your guess: ").strip()

        # int() raises ValueError for text, decimals and empty input
        try:
            guess = int(raw)
        except ValueError:
            print("Please enter a whole number (no letters or decimals).")
            continue

        # Reject numbers outside the allowed range
        if lower_bound <= guess <= upper_bound:
            return guess
        print(f"Your number must be between {lower_bound} and {upper_bound}.")

def hint(answer, current_guess):
    """Tell the player whether their guess was too high or too low."""
    if current_guess > answer:
        print("Too high")
    else:
        print("Too low")

def game_loop():
    """Play one round and update the global statistics."""
    # 'global' is needed because we change these module-level counters
    global games_played, games_won, games_lost, total_guesses, best_game

    answer = random.randint(lower_bound, upper_bound)  # new number every round
    guesses = 0
    won = False

    while guesses < max_attempts:
        # Only valid guesses come back, so only they increase the counter
        current_guess = get_valid_guess(guesses + 1)
        guesses += 1
        total_guesses += 1

        if current_guess == answer:
            won = True
            break

        print("INCORRECT, try again")
        hint(answer, current_guess)

    # ---------- Round result and stats ----------
    games_played += 1
    if won:
        games_won += 1
        print(f"\nCORRECT!!! You got it in {guesses} guess(es).")
        if best_game is None or guesses < best_game:
            best_game = guesses
    else:
        games_lost += 1
        print(f"\nOut of attempts! The number was {answer}.")

def show_stats():
    """Print the player's statistics for the session."""
    print("\n----- Your Statistics -----")
    print(f"Games played: {games_played}")
    print(f"Wins: {games_won}")
    print(f"Losses: {games_lost}")

    # Guard against dividing by zero if no games were played
    if games_played > 0:
        win_rate = games_won / games_played * 100
        average = total_guesses / games_played
        print(f"Win rate: {win_rate:.0f}%")
        print(f"Average guesses per game: {average:.1f}")

    if best_game is not None:
        print(f"Best game: {best_game} guess(es)")
    print("---------------------------")

def play_again():
    """Ask whether to play another round. Returns True for yes."""
    while True:
        choice = input("\nPlay again? (y/n): ").strip().lower()
        if choice in ("y", "yes"):
            return True
        if choice in ("n", "no"):
            return False
        print("Please type 'y' or 'n'.")

def main():
    """Run the game: rules, rounds until the player quits, then stats."""
    rules()
    playing = True
    while playing:
        game_loop()
        playing = play_again()
    show_stats()
    print("Thanks for playing!")

# Only run the game when this file is executed directly
if __name__ == "__main__":
    main()
````
